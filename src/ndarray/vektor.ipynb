```python
import numpy as np


# vektor
vektor = np.array([1, 2, 3])
print("Vektor:", vektor)
```

```python
# Operasi Aritmatika
vektor1 = np.array([1, 2, 3])
vektor2 = np.array([4, 5, 6])
print("Vektor 1:", vektor1)
print("Vektor 2:", vektor2)
print()
print("Penjumlahan:", vektor1 + vektor2)
print("Pengurangan:", vektor1 - vektor2)
print("Perkalian:", vektor1 * vektor2)
print("Pembagian:", vektor1 / vektor2)
```
