```python
import numpy as np

matriks1 = np.array([[1, 2], [3, 4]])
matriks2 = np.array([[5, 6], [7, 8]])
print("Matriks 1:\n", matriks1)
print("Matriks 2:\n", matriks2)
print()
print("Penjumlahan Matriks:\n", matriks1 + matriks2)
print("Pengurangan Matriks:\n", matriks1 - matriks2)
print("Perkalian Matriks:\n", matriks1 * matriks2)
print("Ukuran Matriks 1:", matriks1.size)
print("Ukuran Matriks 2:", matriks2.size)
```
