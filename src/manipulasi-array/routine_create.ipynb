```python
import numpy as np


vektor = np.random.rand(3)
print("Vektor:\n", vektor)

matriks = np.random.rand(2, 3)
print("Matriks:\n", matriks)

tensor = np.random.rand(2, 3, 4)
print("Tensor:\n", tensor)
```

```python
def penjumlahan(a, b):
    return a + b

vektor1 = np.random.rand(3)
vektor2 = np.random.rand(3)
print("Vektor 1:\n", vektor1)
print("Vektor 2:\n", vektor2)
hasil_penjumlahan = penjumlahan(vektor1, vektor2)
print("Hasil Penjumlahan Vektor:\n", hasil_penjumlahan)

# tensor1 = np.random.rand(2, 3, 4)
# tensor2 = np.random.rand(2, 3, 4)
# hasil_penjumlahan = penjumlahan(tensor1, tensor2)
# print("Tensor 1:\n", tensor1)
# print("Tensor 2:\n", tensor2)
# print("Hasil Penjumlahan Tensor:\n", hasil_penjumlahan)
```

```python
# Penjumlahan Vektor + Matriks
vektor = np.random.rand(3)
matriks = np.random.rand(2, 3)
print("Vektor:\n", vektor)
print("Matriks:\n", matriks)
try:
    hasil_penjumlahan = penjumlahan(vektor, matriks)
    print("Hasil Penjumlahan Vektor + Matriks:\n", hasil_penjumlahan)
except ValueError as e:
    print("Error:", e)
```
